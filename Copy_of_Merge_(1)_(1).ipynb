## BOW

```python
def generate_ngrams(text_list, min_n, max_n, stop_words):
    ngrams = {}

    for text in text_list:
        tokens = text.split()
        for n in range(min_n, max_n + 1):
            for i in range(len(tokens) - n + 1):
                ngram = ' '.join(tokens[i:i+n])
                if ngram.split()[0] not in stop_words and ngram.split()[-1] not in stop_words:
                    ngrams[ngram] = ngrams.get(ngram, 0) + 1
    return ngrams

# Assuming clean_text is a list of sentences and stop_words is a list of stop words
ngrams_counts = generate_ngrams(clean_text, 2, 12, stop_words)

sorted_ngrams_counts = sorted(ngrams_counts.items(), key=lambda x: -x[1])

print("Top sorted n-grams and counts overall:")
for ngram, count in sorted_ngrams_counts[:120]:
    print(f"{ngram}: {count}")
```

```python
import re
from sklearn.feature_extraction.text import CountVectorizer


count_vectorizer = CountVectorizer(ngram_range=(2,12), token_pattern=r'(?u)\b\w+\b')

# Fit the vectorizer to the documents and transform the text
count_matrix = count_vectorizer.fit_transform(clean_text)

# Extracting the n-grams and their counts
feature_names = count_vectorizer.get_feature_names_out()

# Summing counts across all documents for each n-gram
overall_counts = count_matrix.sum(axis=0).tolist()[0]

filtered_ngrams_counts = [
    (ngram, count) for ngram, count in zip(feature_names, overall_counts)
    if not (ngram.split()[0] in stop_words or ngram.split()[-1] in stop_words)
]

sorted_ngrams_counts = sorted(filtered_ngrams_counts, key=lambda x: -x[1])

print("Top sorted n-grams and counts overall:")
for ngram, count in sorted_ngrams_counts[:120]:
    print(f"{ngram}: {count}")
```

## Result

```python
# Extracting the top n-grams and consider them for subtraction
top_ngrams = final_ngrams_list2[:120]

# Create a dictionary to store the final n-grams and their counts
final_ngrams_counts = dict(top_ngrams)
# Iterate through the sorted n-grams, starting with the higher n-grams
for i in range(len(top_ngrams)):
    ngram_i, count_i = top_ngrams[i]
    for j in range(i + 1, len(top_ngrams)):
        ngram_j, count_j = top_ngrams[j]
        # Check if the lower n-gram is a complete part of the higher n-gram
        if ngram_i in ngram_j:
            # Subtract the count of the lower n-gram from the higher n-gram
            count_i -= count_j
            if count_i <= 0:
                # Exclude the lower n-gram if its count is zero or negative
                final_ngrams_counts.pop(ngram_i, None)
                break  # No need to continue checking once count is zero or negative
            else:
                final_ngrams_counts[ngram_i] = count_i

# Convert the dictionary back to a sorted list
final_ngrams_list = sorted(final_ngrams_counts.items(), key=lambda x: -x[1])[:80]

# Print the result
print("Top 120 n-grams after subtracting overlapping n-grams:")
for ngram, count in final_ngrams_list:
    print(f"{ngram}: {count}")
```

```python
# Calculating the new values

calculated_values_sample = [(ngram, count, round((count/total_count)*100, 1)) for ngram, count in final_ngrams_list]

# Sorting by the calculated value in descending order
sorted_calculated_values_sample = sorted(calculated_values_sample, key=lambda x: -x[2])

bi_grams = [item for item in calculated_values_sample if len(item[0].split()) == 2]
non_bi_grams = [item for item in calculated_values_sample if len(item[0].split()) != 2]

# Sort non-bi-grams based on count and value
sorted_non_bi_grams = sorted(non_bi_grams, key=lambda x: (-x[1], -x[2]))

# List to store the final sorted n-grams
final_sorted_list4 = []

# Append non-bi-grams and their associated bi-grams
for ngram, count, value in sorted_non_bi_grams:
    final_sorted_list4.append((ngram, count, value))
    for bg in bi_grams:
        if bg[0] in ngram:
            final_sorted_list4.append(bg)
            bi_grams.remove(bg)

# Append the remaining bi-grams
final_sorted_list4 += bi_grams
# Printing the results (ngram, original count, and calculated value)
results_sample = []
for ngram, count, value in final_sorted_list4:
    results_sample.append((ngram, count, value))

results_sample
```
