```python
def count_words(input_string):
    # Split the string into words
    words = input_string.split()
    # Count the number of words
    num_words = len(words)
    return num_words
```

```python
import requests
from bs4 import BeautifulSoup
import html_text

# URL of the webpage you want to extract text from
url = 'https://flatandvilla.com/%D8%A7%D9%84%D9%82%D8%A7%D9%87%D8%B1%D8%A9-%D8%A7%D9%84%D8%AC%D8%AF%D9%8A%D8%AF%D8%A9/%D9%83%D9%85%D8%A8%D9%88%D9%86%D8%AF-%D9%86%D8%A7%D9%8A%D9%84/'

# Fetch the HTML content
response = requests.get(url)
html_content = response.content

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(html_content, 'html.parser')

# Extract text using html_text
extracted_text = html_text.extract_text(
    str(soup),
    guess_layout=True,  # Attempt to guess the layout of the text
    # preserve_formatting=False,  # Preserve the original HTML formatting
    # list_bullets='•',  # Customize list bullet points
    # links='inline',  # Extract links inline with the text
)


# print(extracted_text)
word_count = count_words(extracted_text)
print(f"Number of words: {word_count}")
```

```python
import requests
from readabilipy import simple_json_from_html_string

url = 'https://flatandvilla.com/%D8%A7%D9%84%D9%82%D8%A7%D9%87%D8%B1%D8%A9-%D8%A7%D9%84%D8%AC%D8%AF%D9%8A%D8%AF%D8%A9/%D9%83%D9%85%D8%A8%D9%88%D9%86%D8%AF-%D9%86%D8%A7%D9%8A%D9%84/'
# url="https://www.uvisne.com/project/Track-10-Tower-New-Capital/"
# url="https://suligolfresidence.org/"




headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

response = requests.get(url, headers=headers)

if response.status_code == 200:
    html = response.text
    article = simple_json_from_html_string(html, use_readability=True)

    title = article['title']
    content = article['content']

    print(f"Title: {title}")
    print("Content:")
    # print(content)
    soup = BeautifulSoup(content, 'html.parser')
    text = soup.get_text(separator='\n', strip=True)

    print(text)
else:
    print(f"Failed to retrieve the webpage. Status code: {response.status_code}")
```

```python
import requests
import trafilatura

# url = 'https://flatandvilla.com/%D8%A7%D9%84%D9%82%D8%A7%D9%87%D8%B1%D8%A9-%D8%A7%D9%84%D8%AC%D8%AF%D9%8A%D8%AF%D8%A9/%D9%83%D9%85%D8%A8%D9%88%D9%86%D8%AF-%D9%86%D8%A7%D9%8A%D9%84/'
# url="https://www.uvisne.com/project/Track-10-Tower-New-Capital/"
url="https://suligolfresidence.org/"
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

response = requests.get(url, headers=headers)

if response.status_code == 200:
    html = response.text
    downloaded = trafilatura.fetch_url(url)
    if downloaded is not None:
        result = trafilatura.extract(downloaded)
        if result:
            print(result)
        else:
            print("Failed to extract content.")
    else:
        print("Failed to download the page content.")
else:
    print(f"Failed to retrieve the webpage. Status code: {response.status_code}")
```
